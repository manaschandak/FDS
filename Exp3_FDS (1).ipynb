```python
import pandas as pd

file_path = '/content/panasonic_engineered_sample.xlsx'

try:
    # Load the Excel file into a pandas DataFrame
    df_new = pd.read_excel(file_path)

    # Display the first 30 rows and 5 attributes
    print(f"Dataset shape: {df_new.shape}")
    display(df_new.head(30).iloc[:, :5])
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure it has been uploaded.")
except Exception as e:
    print(f"An error occurred while reading the Excel file: {e}")
```

```python
import numpy as np

first_five_cols_df = df_new.iloc[:, :5]

numeric_cols = first_five_cols_df.select_dtypes(include=np.number).columns

print('Descriptive Statistics for Numeric Columns (First 5 Attributes Only):')
for col in numeric_cols:
    print(f' Column: {col}\n')

    # Frequency
    frequency = df_new[col].count()
    print(f'Frequency: {frequency}')

    # Mean
    mean_val = df_new[col].mean()
    print(f'Mean: {mean_val:.4f}')

    # Mode
    mode_val = df_new[col].mode()
    print(f'Mode: {mode_val.tolist()}')

    # Median
    median_val = df_new[col].median()
    print(f'Median: {median_val:.4f}')

    # Standard Deviation
    std_dev_val = df_new[col].std()
    print(f'Standard Deviation: {std_dev_val:.4f}')

    # Variance
    variance_val = df_new[col].var()
    print(f'Variance: {variance_val:.4f}')

    # Interquartile Range (IQR)
    Q1 = df_new[col].quantile(0.25)
    Q3 = df_new[col].quantile(0.75)
    iqr_val = Q3 - Q1
    print(f'Interquartile Range (IQR): {iqr_val:.4f}')
```

```python
import pandas as pd

df = pd.read_excel('/content/panasonic_engineered_sample.xlsx')

data = df.iloc[:30, 1:6]

for col in data.columns:

    a = []

    for x in data[col]:
        if pd.notna(x):
            a.append(float(x))

    n = len(a)

    total = 0
    for x in a:
        total += x
    mean = total / n

    for i in range(n):
        for j in range(i + 1, n):
            if a[i] > a[j]:
                a[i], a[j] = a[j], a[i]

    if n % 2 == 0:
        median = (a[n//2 - 1] + a[n//2]) / 2
    else:
        median = a[n//2]

    mode = a[0]
    max_count = 0

    for x in a:
        count = 0
        for y in a:
            if x == y:
                count += 1
        if count > max_count:
            max_count = count
            mode = x

    total = 0
    for x in a:
        total += (x - mean) ** 2

    variance = total / n
    sd = variance ** 0.5

    q1_pos = (n - 1) // 4
    q3_pos = 3 * (n - 1) // 4

    q1 = a[q1_pos]
    q3 = a[q3_pos]

    iqr = q3 - q1

    print("\n", col)
    print("Mean: {mean:.3f}")
    print("Median: {median:.3f}")
    print("Mode: {mode:.3f}")
    print("Variance: {variance:.3f}")
    print("Standard Deviation: {sd:.3f}")
    print("Interquartile Range: {iqr:.3f}")
```

```python
import pandas as pd
import math

file_path = '/content/panasonic_engineered_sample.xlsx'

df = pd.read_excel(file_path)

data = df.iloc[:30, 1:6]

for col in data.columns:

    values = []

    for x in data[col]:
        if pd.notna(x):
            values.append(float(x))

    minimum = values[0]
    maximum = values[0]

    for x in values:
        if x < minimum:
            minimum = x
        if x > maximum:
            maximum = x

    width = (maximum - minimum) / 5

    freq = [0, 0, 0, 0, 0]
    mid = []

    for i in range(5):
        lower = minimum + i * width
        upper = minimum + (i + 1) * width

        mid.append((lower + upper) / 2)

        for x in values:
            if i == 4:
                if lower <= x <= upper:
                    freq[i] += 1
            else:
                if lower <= x < upper:
                    freq[i] += 1

    n = 0
    fx = 0

    for i in range(5):
        n += freq[i]
        fx += freq[i] * mid[i]

    mean = fx / n

    cf = []
    total = 0

    for f in freq:
        total += f
        cf.append(total)

    pos = n / 2
    med_group = 0

    for i in range(5):
        if cf[i] >= pos:
            med_group = i
            break

    if med_group == 0:
        prev = 0
    else:
        prev = cf[med_group - 1]

    lower = minimum + med_group * width

    median = lower + ((pos - prev) / freq[med_group]) * width

    mode_group = 0

    for i in range(1, 5):
        if freq[i] > freq[mode_group]:
            mode_group = i

    lower = minimum + mode_group * width

    f0 = freq[mode_group - 1] if mode_group > 0 else 0
    f1 = freq[mode_group]
    f2 = freq[mode_group + 1] if mode_group < 4 else 0

    mode = lower + ((f1 - f0) / (2 * f1 - f0 - f2)) * width

    v = 0

    for i in range(5):
        v += freq[i] * (mid[i] - mean) ** 2

    variance = v / n
    sd = math.sqrt(variance)

    pos = n / 4
    q1_group = 0

    for i in range(5):
        if cf[i] >= pos:
            q1_group = i
            break

    prev = cf[q1_group - 1] if q1_group > 0 else 0
    lower = minimum + q1_group * width

    q1 = lower + ((pos - prev) / freq[q1_group]) * width

    pos = 3 * n / 4
    q3_group = 0

    for i in range(5):
        if cf[i] >= pos:
            q3_group = i
            break

    prev = cf[q3_group - 1] if q3_group > 0 else 0
    lower = minimum + q3_group * width

    q3 = lower + ((pos - prev) / freq[q3_group]) * width

    iqr = q3 - q1

    print("\n", col)
    print("Frequency:", freq)
    print("Grouped Mean:", mean)
    print("Grouped Median:", median)
    print("Grouped Mode:", mode)
    print("Grouped Variance:", variance)
    print("Grouped Standard Deviation:", sd)
    print("Grouped IQR:", iqr)
```
